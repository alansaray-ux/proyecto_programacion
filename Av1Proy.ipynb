```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
 #CREACION DEL GRAFO
 G = nx.Graph()
```

```python
#INGRESO DE ESTACIONES
G.add_node("Plaza_Castilla", linea="L1")
G.add_node("Cuatro_Caminos", linea="L1")
G.add_node("Bilbao", linea="L1")
G.add_node("Tribunal", linea="L1")
G.add_node("Gran_Via", linea="L1")
G.add_node("Sol", linea="L1_L6")# estacion de transbordo L1-L6
G.add_node("Nuevos_Ministerios", linea="L6")
G.add_node("Cuzco", linea="L6")
G.add_node("Moncloa", linea="L6")
G.add_node("Arguelles", linea="L6")
G.add_node("Principe_Pio", linea="L6")
'''
Cada estacion se agrega como un nodo. El valor linea indica a
que linea pertenece.
'''
```

```python
#Conexiones
G.add_edge("Plaza_Castilla", "Cuatro_Caminos", tiempo=3)
G.add_edge("Cuatro_Caminos", "Bilbao", tiempo=2)
G.add_edge("Bilbao", "Tribunal", tiempo=2)
G.add_edge("Tribunal", "Gran_Via", tiempo=1)
G.add_edge("Gran_Via", "Sol", tiempo=2)
 
G.add_edge("Cuatro_Caminos", "Nuevos_Ministerios", tiempo=3)
G.add_edge("Nuevos_Ministerios", "Cuzco", tiempo=2)
G.add_edge("Cuzco", "Moncloa", tiempo=4)
G.add_edge("Moncloa", "Arguelles", tiempo=2)
G.add_edge("Arguelles", "Principe_Pio", tiempo=3)
G.add_edge("Principe_Pio", "Sol", tiempo=5)
#el valor tiempo indica el tiempo de viaje entre estaciones conectadas
```

```python
#estructura basica del grafo
print("ESTRUCTURA DEL GRAFO")
print("Numero total de estaciones (nodos):", G.number_of_nodes())
print("Numero total de conexiones (aristas):", G.number_of_edges())
print("Lista de estaciones:", list(G.nodes()))
print("Lista de conexiones:", list(G.edges()))
```
